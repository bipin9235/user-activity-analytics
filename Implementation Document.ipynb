### **Phase-2: Building Production-Grade Data Pipeline**

### **Assignment 1: Silver Layer Transformation & Data Quality**
* **Create a Silver layer Delta table** with cleaned, validated, and enriched data:
  - Remove duplicates (by timestamp + user_id + action_id)
  - Handle null values and malformed records
  - Add data quality checks (valid timestamps, country code format, required fields)
  - Create a separate **quarantine table** for invalid records
  - Add watermarking for late-arriving data handling
  - Parse and standardize timestamp to UTC
  - Enrich with derived columns (hour_of_day, day_of_week, session indicators)

**Learning goals**: Data quality patterns, schema enforcement, error handling

### **Assignment 2: Gold Layer - Business Metrics & SCD**
* **Build Gold layer aggregate tables**:
  - User activity summary (daily active users, session counts per user)
  - Page performance metrics (avg time on page, bounce rates)
  - Implement **SCD Type 2** for tracking user country changes over time
  - Create pre-aggregated tables for dashboard optimization

**Learning goals**: Dimensional modeling, slowly changing dimensions, performance optimization

### **Assignment 3: Implement Lambda Architecture Comparison**
* **Build both streaming and batch paths**:
  - **Streaming path** (Kappa): Continue your real-time 10-sec ingestion → bronze → silver → gold
  - **Batch path**: Create a scheduled job that reprocesses daily batches for accuracy
  - Compare results between streaming vs batch for the same time window
  - Implement a **merge layer** that reconciles differences

**Learning goals**: Lambda vs Kappa tradeoffs, consistency vs latency, reconciliation patterns

### **Assignment 4: Advanced Streaming Patterns**
* **Implement stateful streaming**:
  - Session windowing (group clicks into user sessions with 30-min timeout)
  - Running aggregations (hourly click counts updated in real-time)
  - Handle out-of-order events with watermarks
  - Implement exactly-once semantics with Delta idempotent writes

**Learning goals**: Stateful stream processing, window functions, event-time vs processing-time

### **Assignment 5: Monitoring, Alerts & Data Quality**
* **Set up Data Quality Monitoring**:
  - Create expectations/assertions on row counts, freshness, null percentages
  - Build a **data health dashboard** showing pipeline lag, failure rates, data quality metrics
  - Set up **alerts** for data freshness (no new data in 15 min) and anomaly detection (sudden drop/spike in events)
  - Implement schema evolution handling with Delta schema enforcement

**Learning goals**: Observability, SLAs, proactive monitoring

### **Assignment 6: Optimization & Scalability**
* **Performance tuning**:
  - Partition your bronze/silver tables by date
  - Implement Z-ordering on frequently filtered columns (user_id, page)
  - Set up auto-compaction and OPTIMIZE schedules
  - Add caching strategies for frequently accessed aggregates
  - Benchmark query performance before/after optimization

**Learning goals**: Delta optimization, partitioning strategies, performance engineering

### **Assignment 7: Disaster Recovery & Debugging**
* **Build resilience**:
  - Implement checkpoint management for streaming jobs
  - Create a backfill mechanism to reprocess historical data
  - Add detailed logging at each layer for debugging
  - Simulate failures (Kafka downtime, schema changes, bad data) and document recovery procedures
  - Version your pipeline code and maintain a rollback strategy

**Learning goals**: Failure modes, recovery patterns, operational excellence

### **Bonus Challenge: Real-time Feature Store**
* Build a feature store for ML use cases:
  - Extract real-time user behavior features (click velocity, favorite pages, active hours)
  - Maintain both online (low-latency lookup) and offline (batch training) feature tables
  - Serve features with < 100ms latency for real-time predictions
